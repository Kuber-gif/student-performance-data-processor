```python
import json
```

```python
with open ("students.json", "r") as file:
    students_raw = json.load(file)
```

```python
def clean_students(students):
    cleaned_students = []
    seen = []
    for student in students:
        if student["name"].strip() == "":
            continue
        if student["marks"] is None:
            continue
        marks = int(student["marks"])

        if marks < 0 or marks > 100:
            continue
        if student["name"] in seen:
            continue

        seen.append(student["name"])
        cleaned_student = {
            "name": student["name"],
            "marks": marks
            }
        cleaned_students.append(cleaned_student)
    
        

    return cleaned_students
```

```python
clean_students(students_raw)
```

```python
students = clean_students(students_raw)
```

```python
def get_passing_students(students):
    passed_student = []
    for student in students:
        if student["marks"] >= 70:
            passed_student.append(student["name"])
    return passed_student
```

```python
def calculate_average(students):
    student_marks = [student["marks"] for student in students]
    avg = sum(student_marks)/len(student_marks)
    return avg
```

```python
def get_top_student(students):
    current_highest = 0 
    name_of_highest = ""
    for student in students:
        if student["marks"] > current_highest:
            current_highest = student["marks"]
            name_of_highest = student["name"]
    return f"The highest score was {current_highest} obtained by {name_of_highest} "
```

```python
def get_result_summary(students):
    passed_students = 0 
    failed_students = 0
    for student in students:
        if student["marks"] >= 70:
            passed_students = passed_students + 1 
        else:
            failed_students = failed_students + 1

    return f"Total No. of Students:{len(students)} , No. of Students passed:{passed_students} , No. of Students failed:{failed_students}"
```

```python
def sort_students_by_marks(students):
    return sorted(students, key=lambda student: student["marks"], reverse=True)
```

```python
get_top_student(students)
```

```python
get_result_summary(students)
```

```python
sort_students_by_marks(students)
```

```python
def generate_report(students):
    passing = get_passing_students(students)
    average = calculate_average(students)
    top = get_top_student(students)
    summary = get_result_summary(students)
    ranking = sort_students_by_marks(students)

    return {
    "passing": passing,
    "average": average,
    "top": top,
    "summary": summary,
    "ranking": ranking
    }
```

```python
generate_report(students)
```
